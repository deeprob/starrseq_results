# STARR-seq peak overlap with open chromatin

Calculate the number and percentage of STARR-seq peaks that overlap existing DNase hypersensitive sites.

```python
from pybedtools import BedTool
```

```python
def overlap_pct(
    a="data/peaks.peak.final.bed",
    b="data/ENCFF285OXK.bed",
    f=1e-9,
):
    """Return overlapping STARR-seq peaks, total peaks, and overlap percentage."""
    A = BedTool(a).sort()
    B = BedTool(b).sort().merge()

    n = A.count()
    k = A.intersect(B, u=True, f=f).count()
    pct = round(100 * k / n, 2) if n else 0.0

    return k, n, pct
```

```python
overlapping_peaks, total_peaks, overlap_percentage = overlap_pct()

print(f"Overlapping STARR-seq peaks: {overlapping_peaks:,}")
print(f"Total STARR-seq peaks: {total_peaks:,}")
print(f"Overlap with DNase hypersensitive sites: {overlap_percentage:.2f}%")
```
